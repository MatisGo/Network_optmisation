Read the csv-file

```python
import csv
from datetime import datetime

prices = {}
date_format = '%d.%m.%Y %H:%M'

with open("Prices.csv", newline="", encoding="utf-8") as f:
    reader = csv.reader(f)              # rows -> lists
    header = next(reader, None)         # optional: grab header
    for row in reader:
        from_time = row[0].split(" - ")[0]
        to_time = row[0].split(" - ")[1]
        date_obj = datetime.strptime(from_time, date_format)
        prices[date_obj] = float(row[1])
        
print(prices)
```
